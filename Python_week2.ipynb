```python
# Rohith Ayitha
# fig02_01.py
"""Comparing integers using if statements and comparison operators."""

print('Enter two integers, and I will tell you',
      'the relationships they satisfy.')

# Read the first integer
number1 = int(input('Enter first integer: '))

# Read the second integer
number2 = int(input('Enter second integer: '))

# Equality / inequality
if number1 == number2:
    print(number1, 'is equal to', number2)
if number1 != number2:
    print(number1, 'is not equal to', number2)

# Less than / greater than
if number1 < number2:
    print(number1, 'is less than', number2)
if number1 > number2:
    print(number1, 'is greater than', number2)

# Less than or equal / greater than or equal
if number1 <= number2:
    print(number1, 'is less than or equal to', number2)
if number1 >= number2:
    print(number1, 'is greater than or equal to', number2)
```

```python
# Rohith Ayitha
# Exercise 2.10
"""Display Sum, average, product, smallest and largest of 3 Integers."""

a = int(input("Enter first integer: "))
b = int(input("Enter second integer: "))
c = int(input("Enter third integer: "))

numbers = [a, b, c]

print("Sum:", sum(numbers))
print("Average:", sum(numbers) / 3)
print("Product:", a * b * c)
print("Smallest:", min(numbers))
print("Largest:", max(numbers))
```

```python
# Rohith Ayitha
# Exercise 2.12
"""Calculate and display the value of a $1000 investment after 10, 20, and 30 years at 7% annual return."""

p = 1000
r = 0.07

for n in [10, 20, 30]:
    a = p * (1 + r) ** n
    print(f"Amount after {n} years: ${a:.2f}")
```
